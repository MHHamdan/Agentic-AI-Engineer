# Solution: preference learning

This solution fits a small Bradley-Terry reward model from deterministic pairwise preferences.

```python
import sys

sys.path.insert(0, "..")

from preferences import TRUE_QUALITY, fit_reward, make_preferences, policy_choice, ranking

preferences = make_preferences()
reward = fit_reward(preferences)

print("true ranking:", ranking(TRUE_QUALITY))
print("reward ranking:", ranking(reward))
print("policy choice:", policy_choice(reward))
```
